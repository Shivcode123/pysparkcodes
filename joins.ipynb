```python
from pyspark.sql import SparkSession 
from pyspark.sql.window import Window
from pyspark.sql.functions import count,row_number,col
spark=SparkSession.builder.appName("join practice").getOrCreate()
employees = [
    (101, "Ravi", 10, 50000),
    (102, "Sita", 20, 60000),
    (103, "John", 10, 55000),
    (104, "Priya", 30, 70000),
    (105, "Arun", 40, 45000),
    (106, "Kiran", 50, 65000)
]
emp_columns = ["emp_id", "emp_name", "dept_id", "salary"]
df1=spark.createDataFrame(employees, emp_columns)
departments = [
    (10, "IT"),
    (20, "HR"),
    (30, "Finance"),
    (40, "Sales"),
    (60, "Marketing")
]
dept_columns = ["dept_id", "dept_name"]
df2=spark.createDataFrame(departments, dept_columns)

joined=df1.join(df2,df1.dept_id==df2.dept_id,"inner")
#joined.show()
window = Window.partitionBy(df2.dept_id).orderBy(df1.salary.desc())

result=joined.withColumn("rn",row_number().over(window))

result.where(result.rn==1).show()




```
