```python
from pyspark.sql import SparkSession 
from pyspark.sql.window import*
from pyspark.sql.functions import*
spark=SparkSession.builder.appName("join practice").getOrCreate()
data = [
    (1, "charan", "IT", 10000),
    (2, "hari", "Finance", 2000),
    (3, "bharath", "HR", 309000),
    (4, "karthik", "Finance", 4567),
    (5, "sanjay", "Finance", 456789),
    (6, "arun", "IT", 3456),
    (7, "vikram", "IT", 45678),
    (8, "rahul", "HR", 34567),
    (9, "ram", "HR", 43567)
]
columns = ["id", "name", "dept", "salary"]
df = spark.createDataFrame(data, columns)
# df.show()
order=Window.partitionBy(col("dept")).orderBy(col("salary").desc())
df=df.withColumn("dense_rank",dense_rank().over(order))\
     .withColumn("Rank",rank().over(order))\
     .withColumn("Row_number",row_number().over(order))
df.show()
  
```
