```python
from pyspark.sql import SparkSession 
from pyspark.sql.window import*
from pyspark.sql.functions import*
spark=SparkSession.builder.appName("join practice").getOrCreate()
data = [
    (1, "raj", 2023, 4000),
    (2, "ramesh", 2023, 5500),
    (3, "rajesh", 2023, 5000),
    (1, "raj", 2024, 4500),
    (2, "ramesh", 2024, 5500),
    (3, "rajesh", 2024, 5200),
    (1, "raj", 2025, 4300),
    (2, "ramesh", 2025, 5300),
    (3, "rajesh", 2025, 5800)
]
columns = ["id", "name", "year", "salary"]
df=spark.createDataFrame(data,columns)
window = Window.partitionBy("name").orderBy(col("year"))
# df1=df.withColumn("last_salary",lag('salary').over(window))\
#     .withColumn("hike",col("salary")-col("last_salary"))\
#     .filter(col("hike")>0)\
#     .groupBy("name").agg(count("name").alias("count"))\
#     .filter(col("count")>0)
# df1.filter(col("count")==2).select("name").show() 
df1=df.withColumn("last_salary",lag('salary').over(window))\
      .withColumn("hike",col("salary")-col("last_salary"))\
      .withColumn("new_hike",lag("hike").over(window))
df1.show()    
df1.filter((col("hike")>0)&(col("new_hike")>0)).select("name").show()

```
