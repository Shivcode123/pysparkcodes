```python
from pyspark.sql import SparkSession 
from pyspark.sql.window import Window
from pyspark.sql.functions import count,row_number,col
spark=SparkSession.builder.appName("join practice").getOrCreate()
data = [
    (1, "f1", "delhi", "hyderabad", 2500),
    (1, "f2", "hyderabad", "kochi", 1700),
    (1, "f3", "kochi", "mangalore", 1800),
    (2, "f1", "mumbai", "ayodhya", 4000),
    (2, "f2", "ayodhya", "chennai", 3000)
]
columns = [
    "cust_id",
    "flight_id",
    "origin",
    "destination",
    "price"
]
df = spark.createDataFrame(data, columns)
#df.show()
result=df.groupBy("cust_id").agg(first("origin").alias("start"),last("destination").alias("end"),sum("price").alias("total_fair"))
result.show()

```
