```python
from pyspark.sql import SparkSession 
from pyspark.sql.window import Window
from pyspark.sql.functions import*
spark=SparkSession.builder.appName("join practice").getOrCreate()
data=[("India", "SL", "India"),
    ("SL", "Aus", "Aus"),
    ("SA", "Eng", "Eng"),
    ("Eng", "NZ", "NZ"),
    ("Aus", "India", "India")
      ]
columns=("Team_1","Team_2","Winner")
df=spark.createDataFrame(data,columns)   
df.show() 
df1=df.select("Team_1")
df2=df.select("Team_2")
#df1.show()
#df2.show()
total=df1.union(df2)
df3=total.groupBy("Team_1").count().withColumnRenamed("Team_1","Team").withColumnRenamed("count","matches_played")
df3.show()
#df.show()
join=df3.join(df,df3.Team==df.Winner,'left').drop("Team_1","team_2")
join.show()
df4=join.groupBy("Team","matches_played").agg(count("Winner").alias("no_of_wins"))
df4=df4.withColumn("no_of_losses",col("matches_played")-col("no_of_wins"))
df4.show()
```
