```python
from pyspark.sql import SparkSession 
from pyspark.sql.window import*
from pyspark.sql.functions import*
spark=SparkSession.builder.appName("join practice").getOrCreate()
data=(['kasireddy naidu'],
      ['konidela ram cheran'],
      ['nandamuri thraraka ramarao'],
      ['charan'])
columns=["name"]
df=spark.createDataFrame(data,columns)
df.show()
df=df.withColumn("parts",split(col("name")," "))
df=df.withColumn("first_name",col("parts")[0])\
     .withColumn("Last_name",when(size("parts")==1,lit("")).otherwise(col("parts")[size("parts")-1]))\
     .withColumn("middle_name",when(size(col("parts"))>2,concat_ws(" ",slice(col("parts"),2,size("parts")-2))).otherwise("") )
df.select("name","first_name","middle_name","Last_name").show()
```
